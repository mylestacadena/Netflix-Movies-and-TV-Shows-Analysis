```python
#Netflix Movies and TV Shows Analysis

#-----Part 1: Data Cleaning and Preprocessing-----

#1. Import necessary libraries and load the dataset
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from IPython.display import display


DATA_PATH = "netflix_titles.csv"

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        "netflix_titles.csv was not found. "
        "Place the dataset in the same folder as this notebook."
    )

df = pd.read_csv(DATA_PATH)

print("Dataset successfully loaded.")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

#2. Identify missing data in the dataset
missing = df.isnull().sum().to_frame(name='Missing Values')
display(missing)

#3. Visualize the missing data using a heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='coolwarm')
plt.title("Missing Data Pattern by Column")
plt.show()

#4. Clean the data
df.drop(columns=['show_id'], inplace=True)  # Drop 'show_id' as it's not useful for analysis

#Fill missing values in 'director' and 'cast' columns with 'Unknown'
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')

#Replace missing values in 'rating' and 'country' with their respective modes or the most frequent values
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
df['country'] = df['country'].fillna(df['country'].mode()[0])

#Fill missing 'date_added' and 'duration' values with their modes
df['date_added'] = df['date_added'].fillna(df['date_added'].mode()[0])
df['duration'] = df['duration'].fillna(df['duration'].mode()[0])

#5. Clean column names and string data
df.columns = df.columns.str.strip()  #Remove leading/trailing whitespace in column names

#Strip extra spaces in all string columns
str_cols = df.select_dtypes(include='object').columns
for col in str_cols:
    df[col] = df[col].str.strip()

#6. Convert 'date_added' column to datetime and extract year and month for analysis
df['date_added'] = pd.to_datetime(df['date_added'])
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month

#7. Re-check for any missing data after cleaning
print("Missing data after cleaning:")
display(df.isnull().sum().to_frame(name='Missing Values'))


# -----Part 2: Feature Engineering-----

#1. Extract numeric duration and its unit from 'duration' column
df['duration_num'] = df['duration'].str.extract(r'(\d+)').astype(float)
df['duration_unit'] = df['duration'].str.extract(r'([a-zA-Z]+)')

#Create a new column for movie durations in minutes: only for movies, not TV shows
df['duration_min'] = df.apply(
    lambda row: row['duration_num'] if row['duration_unit'] == 'min' else None, axis=1)
df['duration_sec'] = df['duration_min'] * 60

#2. Create duration bins for content length grouping
duration_bins = [0, 60, 90, 120, 150, 3000]
bin_labels = ['<1hr', '1-1.5hr', '1.5-2hr', '2-2.5hr', '2.5hr+']
df['duration_bin'] = pd.cut(df['duration_min'], bins=duration_bins, labels=bin_labels)

#3. Extract more features from 'date_added'
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
df['day_added'] = df['date_added'].dt.day
df['weekday_added'] = df['date_added'].dt.day_name()

#4. Encode categorical variables for analysis or modeling
df['type_encoded'] = df['type'].map({'Movie': 0, 'TV Show': 1})
df['rating_encoded'] = df['rating'].astype('category').cat.codes


#-----Part 3: Descriptive Statistics and Exploration-----

#Display descriptive statistics for all columns
print("\nDescriptive Statistics:")
display(df.describe(include='all'))

#Display top 10 most frequent content ratings
print("\nTop 10 Ratings:")
top_ratings_df = df['rating'].value_counts().head(10).reset_index()
top_ratings_df.columns = ['Rating', 'Count']
top_ratings_df.index += 1
display(top_ratings_df)

#Show distribution of content types (Movies vs TV Shows)
print("\nContent Type Distribution:")
content_type = df['type'].value_counts().reset_index()
content_type.columns = ['Type', 'Count']
content_type.index += 1
display(content_type)

#Show number of releases per year
print("\nContent Released Per Year:")
year_rel = df['release_year'].value_counts().sort_index().reset_index()
year_rel.columns = ['Release Year', 'Count']
year_rel.index += 1
display(year_rel)


#-----Part 4: Visualizations-----

#Top 10 countries producing Netflix content using countplot
df['country'] = df['country'].str.split(', ')
df_exp = df.explode('country')
top_countries = df_exp['country'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.countplot(data=df_exp[df_exp['country'].isin(top_countries.index)],
              y='country',
              order=top_countries.index,
              palette='viridis')
plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()

#Distribution of top 10 content ratings using barplot
top_ratings = df['rating'].value_counts().head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=top_ratings.index, y=top_ratings.values, palette='pastel')
plt.title('Distribution of Netflix Content Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()

#Movies vs. TV Shows released per year using scatterplot
df_year_type = df.groupby(['release_year', 'type']).size().reset_index(name='count')
plt.figure(figsize=(12,6))
sns.scatterplot(data=df_year_type, x='release_year', y='count', hue='type', style='type', palette='tab10')
plt.title('Number of Movies vs TV Shows Released Each Year')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.show()

#Growth trend of content types over the years using violin plot
plt.figure(figsize=(10,6))
sns.violinplot(data=df,
               x='type',
               y='release_year',
               inner='quartile',
               palette='Set2',
               cut=0)
plt.title('Growth Trend of Netflix Content by Type Over the Years')
plt.xlabel('Type')
plt.ylabel('Release Year')
plt.tight_layout()
plt.show()


#Rating Distribution by Genre

#Preprocess 'listed_in' column by exploding multiple genres
df_genre = df.copy()
df_genre['listed_in'] = df_genre['listed_in'].str.split(', ')
df_genre = df_genre.explode('listed_in')

#Remove invalid 'rating' entries mistakenly stored as durations
invalid_ratings = ['74 min', '84 min', '66 min']
df_genre = df_genre[~df_genre['rating'].isin(invalid_ratings)]

#Retain genres with sufficient sample size: >30 entries
genre_counts = df_genre['listed_in'].value_counts()
valid_genres = genre_counts[genre_counts > 30].index
df_genre = df_genre[df_genre['listed_in'].isin(valid_genres)]

#Rating distribution by genre using FacetGrid
g2 = sns.FacetGrid(df_genre, col='listed_in', col_wrap=3, height=4, sharex=False, sharey=False)
g2.map(sns.countplot, 'rating', order=df_genre['rating'].value_counts().index, palette='coolwarm')

#Rotate x-axis labels
for ax in g2.axes.flatten():
    ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')

#Add axis titles and overall title
g2.set_titles('{col_name}')
g2.set_axis_labels("Rating", "Count")
g2.fig.suptitle("Rating Distribution by Genre", y=1.05)
plt.tight_layout()
plt.show()
```

```python
#World Happiness Report

#-----Part 1: Data Cleaning and Preprocessing-----

#1. Import necessary libraries and load the dataset
from google.colab import files                               #For uploading files
import pandas as pd                                          #For data manipulation
import numpy as np                                           #For numerical operations
import matplotlib.pyplot as plt                              #For visualization with Matplotlib
import seaborn as sns                                        #For visualization with Seaborn
from sklearn.preprocessing import LabelEncoder,MinMaxScaler  #Preprocessing tools
from IPython.display import display                          #For displaying dataframes
import plotly.graph_objects as go                            #For interactive visualizations

#Upload and read the dataset
uploaded = files.upload()
df = pd.read_csv("world_happiness_report.csv")

#2. Clean column names: strip whitespace and replace spaces with underscores
df.columns = df.columns.str.strip().str.replace(" ", "_")

#3. View column names for inspection
print("Column Names:")
print(df.columns.tolist())

#4. Check for missing values
missing = df.isnull().sum().to_frame(name='Missing Values')
display(missing)

#5. Visualize missing data patterns using a heatmap
plt.figure(figsize=(14, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='coolwarm')
plt.title("Missing Data Pattern by Column")
plt.show()

#6. Fill missing values in numerical columns using mean
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

#7. Fill missing values in object columns using mode
object_cols = df.select_dtypes(include='object').columns
for col in object_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

#8. Convert specific columns to numeric in case of type inconsistencies
df['Happiness_Score'] = pd.to_numeric(df['Happiness_Score'], errors='coerce')
if 'Mental_Health_Index' in df.columns:
    df['Mental_Health_Index'] = pd.to_numeric(df['Mental_Health_Index'], errors='coerce')

#9. Fill any remaining missing values in key columns
df['Happiness_Score'] = df['Happiness_Score'].fillna(df['Happiness_Score'].mean())
if 'Mental_Health_Index' in df.columns:
    df['Mental_Health_Index'] = df['Mental_Health_Index'].fillna(df['Mental_Health_Index'].mean())

#10. Final check for missing data after cleaning
print("\nMissing data after cleaning:")
display(df.isnull().sum().to_frame(name='Missing Values'))


#-----Part 2: Feature Engineering-----

#1. Bin Happiness Score into categorical groups
score_bins = [0, 3, 5, 6.5, 8]
score_labels = ['Low', 'Medium', 'High', 'Very High']
df['Happiness_Category'] = pd.cut(df['Happiness_Score'], bins=score_bins, labels=score_labels)

#2. Bin GDP per Capita into categories
gdp_bins = [0, 0.5, 1.0, 1.5, 2.0, df['GDP_per_Capita'].max()]
gdp_labels = ['Low GDP', 'Medium GDP', 'High GDP', 'Very High GDP', 'Extreme GDP']
df['GDP_Category'] = pd.cut(df['GDP_per_Capita'], bins=gdp_bins, labels=gdp_labels, include_lowest=True)

#3. Encode country names into numeric labels for modeling
df['Country_Encoded'] = LabelEncoder().fit_transform(df['Country'])


# ----- Part 3: Descriptive Statistics and Exploration -----

#1. Display descriptive statistics for all columns
print("\nDescriptive Statistics:")
display(df.describe(include='all'))

#2. Identify top 10 countries with the most entries
print("\nTop 10 Countries by Record Count:")
country_counts = df['Country'].value_counts().head(10).reset_index()
country_counts.columns = ['Country', 'Count']
country_counts.index += 1
display(country_counts)

#3. View basic distribution of Happiness Score
print("\nHappiness Score Distribution:")
display(df['Happiness_Score'].describe())

#4. Check number of records per year
print("\nRecords Per Year:")
year_counts = df['Year'].value_counts().sort_index().reset_index()
year_counts.columns = ['Year', 'Count']
year_counts.index += 1
display(year_counts)


# ----- Part 4: Visualizations -----

#Top 10 happiest countries in a selected year using barplot
year = 2023
df_year = df[df['Year'] == year]
top_countries = df_year.groupby('Country')['Happiness_Score'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_countries.values, y=top_countries.index, palette="viridis")
plt.title(f"Top 10 Happiest Countries in {year}")
plt.xlabel("Average Happiness Score")
plt.xlim(0, 10)
plt.ylabel("Country")
plt.show()

#Distribution of happiness scores using Histogram with KDE
plt.figure(figsize=(10, 5))
sns.histplot(df['Happiness_Score'], bins=20, kde=True, color='skyblue')
plt.title("Distribution of Happiness Scores")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")
plt.show()

#Number of records per year using countplot
plt.figure(figsize=(10, 5))
sns.countplot(x='Year', data=df, palette='pastel')
plt.title("Number of Records per Year")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()

#Normalized trends of happiness, GDP, and mental health over time using line plot

#1. Prepare and normalize time series data
trend_data = df.groupby('Year')[['Happiness_Score', 'GDP_per_Capita', 'Mental_Health_Index']].mean().reset_index()
scaler = MinMaxScaler()
normalized = scaler.fit_transform(trend_data[['Happiness_Score', 'GDP_per_Capita', 'Mental_Health_Index']])
normalized_df = pd.DataFrame(normalized, columns=['Happiness_Score', 'GDP_per_Capita', 'Mental_Health_Index'])
normalized_df['Year'] = trend_data['Year']

#2. Plot the trends
plt.figure(figsize=(12, 6))
plt.plot(normalized_df['Year'], normalized_df['Happiness_Score'], marker='o', label='Happiness Score')
plt.plot(normalized_df['Year'], normalized_df['GDP_per_Capita'], marker='o', label='GDP per Capita')
plt.plot(normalized_df['Year'], normalized_df['Mental_Health_Index'], marker='o', label='Mental Health Index')
plt.title("Trends Over Time: Happiness Score, GDP per Capita, and Mental Health Index")
plt.xlabel("Year")
plt.ylabel("Normalized Value (0 to 1)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

#Using Plotly

#1. Get top 10 happiest countries overall based on average score
top10_countries = (
    df.groupby("Country")["Happiness_Score"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .index
)
top10_df = df[df["Country"].isin(top10_countries)]
years = sorted(top10_df['Year'].unique())

#2. Get global min and max for fixed color scale
min_score = top10_df['Happiness_Score'].min()
max_score = top10_df['Happiness_Score'].max()

#3. Create Plotly scatter geo map per year
fig = go.Figure()

for year in years:
    year_data = top10_df[top10_df['Year'] == year]
    fig.add_trace(go.Scattergeo(
        locations=year_data['Country'],
        locationmode='country names',
        marker=dict(
            size=year_data['Happiness_Score'] * 5,
            color=year_data['Happiness_Score'],
            colorscale='Viridis',
            coloraxis='coloraxis'
        ),
        hovertemplate="<b>%{text}</b><br>Happiness Score: %{customdata[0]:.2f}<br>GDP per Capita: %{customdata[1]:,.0f}<extra></extra>",
        text=year_data['Country'],
        customdata=year_data[['Happiness_Score', 'GDP_per_Capita']].values,
        name=str(year),
        visible=(year == years[0])
    ))

#Configure layout, dropdown, and color scale
fig.update_layout(
    title='Top 10 Happiest Countries (by Year)',
    geo=dict(projection_type='natural earth'),
    coloraxis=dict(
        colorscale='Viridis',
        cmin=min_score,
        cmax=max_score,
        colorbar=dict(title="Happiness Score")
    ),
    updatemenus=[{
        "buttons": [
            {
                "label": str(year),
                "method": "update",
                "args": [
                    {"visible": [y == year for y in years]},
                    {"title": f"Top 10 Happiest Countries in {year}"}
                ]
            }
            for year in years
        ],
        "direction": "down",
        "showactive": True,
        "x": 0.13,
        "xanchor": "left",
        "y": 1.1,
        "yanchor": "top"
    }],
    annotations=[
        dict(
            text="Year:",
            showarrow=False,
            x=0.05,
            xanchor="left",
            y=1.1,
            yanchor="top",
            xref="paper",
            yref="paper",
            font=dict(size=14)
        )
    ]
)

fig.show()
```

```python
#Students Performance in Exams

#-----Part 1: Data Cleaning and Preprocessing-----

#1. Import necessary libraries and load the dataset
from google.colab import files                               #For uploading files
import pandas as pd                                          #For data manipulation
import numpy as np                                           #For numerical operations
import matplotlib.pyplot as plt                              #For visualization with Matplotlib
import seaborn as sns                                        #For visualization with Seaborn
from sklearn.preprocessing import LabelEncoder               #Preprocessing tools
from IPython.display import display                          #For displaying dataframes
import plotly.graph_objects as go                            #For interactive visualizations

#Upload and read the dataset
uploaded = files.upload()
df = pd.read_csv("exams.csv")

#2. Identify missing data in the dataset
missing = df.isnull().sum().to_frame(name='Missing Values')
display(missing)

#3. Visualize the missing data using a heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='coolwarm')
plt.title("Missing Data Pattern by Column")
plt.show()

#4.Strip any leading/trailing whitespace from column names and string values

# Strip whitespace from column names
df.columns = df.columns.str.strip()

#Strip whitespace from all string entries in object-type columns
str_cols = df.select_dtypes(include='object').columns
for col in str_cols:
    df[col] = df[col].str.strip()

#5. Check missing data after cleaning
print("Missing data after cleaning:")
display(df.isnull().sum().to_frame(name='Missing Values'))

#6. Handle missing numeric data by filling with median
for col in ['math score', 'reading score', 'writing score']:
    df[col].fillna(df[col].median(), inplace=True)

#-----Part 2: Feature Engineering-----

#1. Encode categorical columns into numerical format using LabelEncoder
le = LabelEncoder()
categorical_cols = ['gender', 'race/ethnicity', 'parental level of education',
                    'lunch', 'test preparation course']

#Apply label encoding and create new encoded columns
for col in categorical_cols:
    df[col + '_encoded'] = le.fit_transform(df[col])

#2. Bin exam scores into letter grades
score_bins = [0, 60, 70, 80, 90, 100]
score_labels = ['F', 'D', 'C', 'B', 'A']

#Apply binning to each subject's score
for subject in ['math score', 'reading score', 'writing score']:
    df[subject + '_grade'] = pd.cut(df[subject], bins=score_bins, labels=score_labels, right=True)

#3. Calculate the average score across the three subjects for each student
df['average_score'] = df[['math score', 'reading score', 'writing score']].mean(axis=1)

#-----Part 3: Descriptive Statistics and Exploration-----

#1. Display descriptive statistics to understand central tendencies and distribution of data
print("\nDescriptive Statistics:")
display(df.describe(include='all'))

#Count of students by gender
print("\nGender Distribution:")
display(df['gender'].value_counts().reset_index().rename(columns={'index': 'Gender', 'gender': 'Count'}))

#Count of students based on parental education level
print("\nParental Education Level Distribution:")
display(df['parental level of education'].value_counts().reset_index().rename(columns={'index': 'Parental Education', 'parental level of education': 'Count'}))


#-----Part 4: Visualizations-----

#Average scores based on parental education level using barplot
top_education = df.groupby('parental level of education')['average_score'].mean().sort_values(ascending=False).reset_index()
display(top_education)

plt.figure(figsize=(10, 6))
sns.barplot(data=top_education, x='average_score', y='parental level of education', palette='coolwarm')
plt.title("Average Student Scores by Parental Level of Education", fontsize=14)
plt.xlabel("Average Score", fontsize=12)
plt.ylabel("Parental Level of Education", fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

#Average scores by race/ethnicity group using barplot
top_ethnic = df.groupby('race/ethnicity')['average_score'].mean().sort_values(ascending=False).reset_index()
display(top_ethnic)

plt.figure(figsize=(8, 5))
sns.barplot(data=top_ethnic, x='average_score', y='race/ethnicity', palette='coolwarm')
plt.title("Average Student Scores by Race/Ethnicity Group", fontsize=14)
plt.xlabel("Average Score", fontsize=12)
plt.ylabel("Race/Ethnicity", fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

#Comparison of average scores between male and female students using barplot
top_gender = df.groupby('gender')['average_score'].mean().sort_values(ascending=False).reset_index()
display(top_gender)

plt.figure(figsize=(6, 5))
sns.barplot(data=top_gender, x='gender', y='average_score', palette='coolwarm')
plt.title("Average Student Scores by Gender", fontsize=14)
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Average Score", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

#Distribution of students' average scores using histogram with KDE
plt.figure(figsize=(10, 6))
sns.histplot(df['average_score'], bins=20, kde=True, color='skyblue', edgecolor='black')
plt.title("Distribution of Students' Average Scores (Math, Reading, Writing)", fontsize=14)
plt.xlabel("Average Score (Out of 100)", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()


#Score distribution by parental education level using violin plot with plotly

#1. Define score types
score_types = ['math score', 'reading score', 'writing score']

#2. Create a violin plot for each subject to show score distribution by parental education level
fig = go.Figure()

for i, score in enumerate(score_types):
    fig.add_trace(go.Violin(
        x=df['parental level of education'],
        y=df[score],
        name=score,
        box_visible=True,
        meanline_visible=True,
        points="all",
        visible=(i == 0)
    ))

#Add dropdown menu to switch between subjects
dropdown_buttons = [
    dict(label=score,
         method="update",
         args=[
             {"visible": [i == j for j in range(len(score_types))]},
             {"yaxis": {
                 "title": score,
                 "range": [0, 120]
              },
              "title": f"{score.capitalize()} Distribution by Parental Level of Education"}
         ])
    for i, score in enumerate(score_types)
]

#Final layout
fig.update_layout(
    title="Math Score Distribution by Parental Level of Education",
    xaxis_title="Parental Level of Education",
    yaxis=dict(title="Math Score", range=[0, 120]),
    xaxis_tickangle=45,
    updatemenus=[dict(
        type="dropdown",
        direction="down",
        pad={"r": 10, "t": 10},
        showactive=True,
        active=0,
        buttons=dropdown_buttons,
        x=0.0,
        xanchor="left",
        y=1.15,
        yanchor="top"
    )],
    height=600
)


fig.show()

#Exam Scores, Demographics, and Socioeconomic Factors in Student Performance using correlation heatmap with plotly

#1. Duplicate dataset and encode relevant features
df_encoded = df.copy()
df_encoded['gender'] = df_encoded['gender'].map({'female': 0, 'male': 1})
df_encoded['lunch'] = df_encoded['lunch'].map({'free/reduced': 0, 'standard': 1})
df_encoded['test preparation course'] = df_encoded['test preparation course'].map({'none': 0, 'completed': 1})

#2. Assign order to parental education levels for ordinal encoding
education_order = [
    "some high school",
    "high school",
    "some college",
    "associate's degree",
    "bachelor's degree",
    "master's degree"
]

#3. Encode parental education and race/ethnicity
df_encoded['parental level of education'] = pd.Categorical(
    df['parental level of education'],
    categories=education_order,
    ordered=True
).codes

df_encoded['race/ethnicity'] = df['race/ethnicity'].astype('category').cat.codes

#4. Define variable groups
scores_vars = ['math score', 'reading score', 'writing score']
# Scores and demographics
scores_demo_vars = scores_vars + ['gender', 'lunch', 'test preparation course']
# Scores, demographics, parental education, and race
all_vars = scores_demo_vars + ['parental level of education', 'race/ethnicity']

#5. Compute correlation matrices
corr_scores = df_encoded[scores_vars].corr()
corr_scores_demo = df_encoded[scores_demo_vars].corr()
corr_all = df_encoded[all_vars].corr()

#6. Define variable groups
variable_sets = {
    'Exam scores only': ['math score', 'reading score', 'writing score', 'average_score'],
    'Scores and demographics': ['math score', 'reading score', 'writing score', 'average_score', 'gender', 'lunch', 'test preparation course'],
}
fig = go.Figure()

#Heatmap for scores only (
fig.add_trace(go.Heatmap(
    z=corr_scores.values,
    x=corr_scores.columns,
    y=corr_scores.index,
    colorscale='Spectral',
    colorbar=dict(title='Correlation'),
    zmin=-1, zmax=1,
    visible=True,
    name='Exam scores only'
))

#Heatmap for scores and demographics
fig.add_trace(go.Heatmap(
    z=corr_scores_demo.values,
    x=corr_scores_demo.columns,
    y=corr_scores_demo.index,
    colorscale='Spectral',
    colorbar=dict(title='Correlation'),
    zmin=-1, zmax=1,
    visible=False,
    name='Scores and demographics'
))

#Heatmap for all variables
fig.add_trace(go.Heatmap(
    z=corr_all.values,
    x=corr_all.columns,
    y=corr_all.index,
    colorscale='Spectral',
    colorbar=dict(title='Correlation'),
    zmin=-1, zmax=1,
    visible=False,
    name='All variables'
))

#Final layout
fig.update_layout(
    title="Comparison of Correlation Patterns Across Exam Scores, Demographics, and Socioeconomic Factors in Student Performance",
    updatemenus=[
        dict(
            buttons=list([
                dict(label="Exam scores only",
                     method="update",
                     args=[{"visible": [True, False, False]},
                           {"title": "Correlation Matrix: Exam scores only"}]),
                dict(label="Scores and demographics",
                     method="update",
                     args=[{"visible": [False, True, False]},
                           {"title": "Correlation Matrix: Scores and demographics"}]),
                dict(label="All variables",
                     method="update",
                     args=[{"visible": [False, False, True]},
                           {"title": "Correlation Matrix: All variables"}]),
            ]),
            direction="down",
            pad={"r": 10, "t": 10},
            showactive=True,
            x=0,
            xanchor="left",
            y=1.15,
            yanchor="top"
        ),
    ]
)

fig.show()
```

```python
#Global CO2 Emissions

#-----Part 1: Data Cleaning and Preprocessing-----

#1. Import necessary libraries and load the dataset
from google.colab import files                               #For uploading files
import pandas as pd                                          #For data manipulation
import numpy as np                                           #For numerical operations
import matplotlib.pyplot as plt                              #For visualization with Matplotlib
import seaborn as sns                                        #For visualization with Seaborn
from sklearn.preprocessing import LabelEncoder               #Preprocessing tools
from IPython.display import display                          #For displaying dataframes
import plotly.graph_objects as go                            #For interactive visualizations
import plotly.colors as pc

#Upload and read the dataset
uploaded = files.upload()
df = pd.read_csv("co2_emissions.csv")  # adjust filename if needed

#3. Identify missing data in the dataset
missing = df.isnull().sum().to_frame(name='Missing Values')
display(missing)

#4. Visualize the missing data using a heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='coolwarm')
plt.title("Missing Data Pattern by Column")
plt.show()

#5. Drop unnecessary identifier column
if 'ObjectId' in df.columns:
    df.drop(columns=['ObjectId'], inplace=True)

#6. Fill missing numeric values with median, categorical with mode
numeric_cols = df.select_dtypes(include=[np.number]).columns
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

object_cols = df.select_dtypes(include='object').columns
for col in object_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

#7. Clean string columns: strip extra whitespace from column names and string entries
df.columns = df.columns.str.strip()
for col in object_cols:
    df[col] = df[col].str.strip()

#8. Bin CO2 emissions for the year 1995 into intervals
if 'F1995' in df.columns:
    bins = [0, 1, 5, 10, 20, 50]
    labels = ['0-1', '1-5', '5-10', '10-20', '20-50']
    df['CO2_emission_bins'] = pd.cut(df['F1995'], bins=bins, labels=labels)

#9. Check how many missing values exist in the 'Country' column
missing_countries = df['Country'].isnull().sum()
print(f"Missing values in 'Country': {missing_countries}")

#10.Final missing check
print("Missing data after cleaning:")
display(df.isnull().sum().to_frame(name='Missing Values'))

#11. Reshape data to long format (tidy data): one row per country-year pair
year_columns = [col for col in df.columns if col.startswith('F') and col[1:].isdigit()]
df_long = df.melt(
    id_vars=[col for col in df.columns if col not in year_columns],
    value_vars=year_columns,
    var_name='Year',
    value_name='CO2 Emissions'
)
df_long['Year'] = df_long['Year'].str[1:].astype(int)


#-----Part 2: Feature Engineering-----

#1. Categorize emission levels into bins based on value range
df_long['CO2_bin'] = pd.cut(df_long['CO2 Emissions'],
                            bins=[0, 100, 500, 1000, 5000, np.inf],
                            labels=['Very Low', 'Low', 'Moderate', 'High', 'Very High'])

#2. Calculate CO2 emissions per capita
if 'Population' in df_long.columns:
    df_long['CO2_per_capita'] = df_long['CO2 Emissions'] / df_long['Population']

#3. Encode country names
if 'Country' in df_long.columns:
    le = LabelEncoder()
    df_long['Country_encoded'] = le.fit_transform(df_long['Country'])


#-----Part 3: Descriptive Statistics and Exploration-----

#1. Display descriptive statistics for all columns
print("\nDescriptive Statistics:")
display(df_long.describe(include='all'))

#2. Clean up 'Country' column and display top countries
if 'Country' in df.columns:
    df['Country'] = df['Country'].str.strip().str.title()
    print("\nTop 10 Countries in CO2 Emission:")
    display(df['Country'].value_counts().head(10))


#-----Part 4: Visualization-----

#1. Top 10 Countries by Mean CO₂ Emissions in 2018 using barplot
if 'F2018' in df.columns:
    top_emitters_2018 = df.groupby('Country')['F2018'].mean().sort_values(ascending=False).head(10)

    plt.figure(figsize=(12, 6))
    sns.barplot(x=top_emitters_2018.values, y=top_emitters_2018.index, palette='OrRd')
    plt.title('Top 10 Countries by Mean CO₂ Emissions in 2018')
    plt.xlabel('Mean CO₂ Emissions (2018)')
    plt.ylabel('Country')
    plt.grid(True, axis='x')
    plt.tight_layout()
    plt.show()
else:
    print("Column 'F2018' not found in the dataset.")

#2. Count of Data Entries per Year using countplot
if 'Year' in df_long.columns:
    plt.figure(figsize=(12, 6))
    sns.countplot(x='Year', data=df_long, palette='coolwarm')
    plt.title('CO₂ Emission Data Entries per Year')
    plt.xticks(rotation=45)
    plt.ylabel('Number of Entries')
    plt.tight_layout()
    plt.show()

#3. Distribution of CO2 emissions per country using violin plot with plotly

#1. Extract year columns and numeric values
year_columns = [col for col in df.columns if col.startswith('F') and col[1:].isdigit()]
years = sorted([int(col[1:]) for col in year_columns])

#2. Define color palette and dictionary for country color mapping
color_palette = pc.qualitative.Set2
max_colors = len(color_palette)

#3. Prepare figure and dropdown menu
fig = go.Figure()
dropdown_buttons = []
country_colors = {}

#4. Create violin plots for each year’s top 10 countries
for i, year in enumerate(years):
    year_col = f'F{year}'
    # Top 10 countries by mean CO₂ emissions for that year
    top10 = df.groupby('Country')[year_col].mean().sort_values(ascending=False).head(10).index.tolist()
    df_top10 = df[df['Country'].isin(top10)]

    for k, country in enumerate(top10):
        country_data = df_top10[df_top10['Country'] == country][year_col].dropna()

        if country not in country_colors:
            country_colors[country] = color_palette[k % max_colors]

        fig.add_trace(go.Violin(
            x=[country] * len(country_data),
            y=country_data,
            name=country,
            visible=(i == 0),
            box_visible=True,
            meanline_visible=True,
            line_color=country_colors[country],
            points='all',
            legendgroup=country,
            showlegend=True,
            hoveron='points+kde',
        ))

    #Visibility for dropdown
    visibility = [False] * (len(years) * 10)
    for j in range(i * 10, (i + 1) * 10):
        visibility[j] = True

    dropdown_buttons.append(dict(
        label=str(year),
        method='update',
        args=[
            {"visible": visibility},
            {"title": f"Distribution of CO₂ Emissions in {year} (Top 10 Countries)",
             "xaxis.title": "Country",
             "yaxis.title": "CO₂ Emissions (Metric Tons)"}
        ]
    ))

#5. Final layout
fig.update_layout(
    title=f"Distribution of CO₂ Emissions in {years[0]} (Top 10 Countries)",
    xaxis_title="Country",
    yaxis_title="CO₂ Emissions (Metric Tons)",
    updatemenus=[dict(
        buttons=dropdown_buttons,
        direction="down",
        pad={"r": 10, "t": 10},
        showactive=True,
        x=0.12,
        xanchor="left",
        y=1.15,
        yanchor="top"
    )],
    legend_title="Country",
    legend=dict(itemsizing='trace'),
    annotations=[
        dict(
            text="Year:",
            showarrow=False,
            x=0.05,
            xanchor="left",
            y=1.1,
            yanchor="top",
            xref="paper",
            yref="paper",
            font=dict(size=14)
        )
    ]
)

fig.show()

#4. Trends over time using line plot

#1. Prepare list of countries and years
all_countries = df['Country'].dropna().unique()
year_columns = [col for col in df.columns if col.startswith('F') and col[1:].isdigit()]
years = [int(col[1:]) for col in year_columns]

fig = go.Figure()

#2. Create a trace for each country
for country in sorted(all_countries):
    country_data = df[df['Country'] == country]
    emissions = country_data[year_columns].mean().values

    fig.add_trace(go.Scatter(
        x=years,
        y=emissions,
        mode='lines+markers',
        name=country,
        visible=False
    ))

fig.data[0].visible = True
default_country = sorted(all_countries)[0]

#3. Create dropdown menu to select country
dropdown_buttons = []
for i, country in enumerate(sorted(all_countries)):
    visibility = [False] * len(all_countries)
    visibility[i] = True
    dropdown_buttons.append(dict(
        label=country,
        method='update',
        args=[
            {'visible': visibility},
            {'title': f'CO₂ Emissions Trend (1995–2018): {country}'}
        ]
    ))

#4. Final layout
fig.update_layout(
    title=f'CO₂ Emissions Trend (1995–2018): {default_country}',
    xaxis=dict(title='Year', tickmode='linear', dtick=1),
    yaxis_title='Mean CO₂ Emissions (Metric Tons)',
    updatemenus=[dict(
        buttons=dropdown_buttons,
        direction='down',
        pad={"r": 10, "t": 10},
        showactive=True,
        x=0.15,
        xanchor='left',
        y=1.13,
        yanchor='top'
    )],
    annotations=[
        dict(
        text="Country:",
        showarrow=False,
        x=0.05,
        xanchor="left",
        y=1.1,
        yanchor="top",
        xref="paper",
        yref="paper",
        font=dict(size=14)
    )]
)

fig.show()

```
